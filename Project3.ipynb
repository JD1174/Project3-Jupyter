# Relational Database Analysis: Querying the Chinook Database

The objective of this notebook is to connect to a relational database system, extract customer purchase records, and transform them into a clean, structured pandas DataFrame. We will be using the SQLite version of the Chinook database, which represents a digital media store with tables for artists, albums, media tracks, invoices, and customers.

### Database Setup and Extraction

I will download the SQLite version of the Chinook database directly and establish a database connection. Then, I will write a SQL query to join the `Customer`, `Invoice`, `InvoiceLine`, `Track`, and `Album` tables to pull together the requested columns: Customer LastName, FirstName, Track Name, and Album Title.

```python
import urllib.request
import sqlite3
import pandas as pd

# Download the Chinook SQLite database file from a reliable public mirror
url = 'https://raw.githubusercontent.com/lerocha/chinook-database/master/ChinookDatabase/DataSources/Chinook_Sqlite.sqlite'
db_file = 'Chinook.sqlite'
urllib.request.urlretrieve(url, db_file)

# Connect to the SQLite database
conn = sqlite3.connect(db_file)

# Define the SQL query to join customer purchases
query = """
SELECT
    c.LastName,
    c.FirstName,
    t.Name AS Name,
    a.Title AS Title
FROM Customer c
JOIN Invoice i ON c.CustomerId = i.CustomerId
JOIN InvoiceLine il ON i.InvoiceId = il.InvoiceId
JOIN Track t ON il.TrackId = t.TrackId
JOIN Album a ON t.AlbumId = a.AlbumId
ORDER BY c.LastName ASC, c.FirstName ASC;
"""

# Read the SQL query results into a pandas DataFrame
df_chinook = pd.read_sql_query(query, conn)

# Close the connection
conn.close()

# Display the first five rows of our results to match the assignment specifications
display(df_chinook.head(5))
```

### Query Analysis and Conclusions

* **Relational Joins**: By joining the core operational tables, we successfully traced each customer's specific item purchases back to their originating albums.
* **Sorted Output**: The resulting DataFrame is properly structured and sorted alphabetically by Customer LastName and FirstName, matching the expected output layout.
